```python
import requests 
import time
import threading
import os
```

```python
def download_file(process_name, url, file_path):
    try:
        print(F"Download Process name started: {process_name}")
        response = requests.get(url)
        print(response)
        with open(file_path, "wb") as file:
            for chunk in response.iter_content(chunk_size =  8192):
                if chunk:
                    file.write(chunk)
        print("File downloaded Successfully")
    except Exception as e:
        print("Error Downloading File : {e}")
    print(f"Process name completed : {process_name}")
```

```python
stop_signal = threading.Event()  ## bydeafault is off

def watch_file(file_path):
    print(f"Watching {file_path}")
    while not stop_signal.is_set():
        print("Watching file ...")
        time.sleep(2)
    print("Watch exiting clearly ...")

# file_path =
t1_Manually= threading.Thread(target=watch_file, args=("a.txt",))
t1_Manually.start()

stop_signal.set() ## Event value to be true

t1_Manually.join()
time.sleep(5)

print("Main Program exited clearly.")
```

```python
stop_signal_1 = threading.Event()  ## bydeafault is off
stop_signal_2 = threading.Event()  ## bydeafault is off
stop_signal_3 = threading.Event()  ## bydeafault is off

def watch_file(thread_id,file_path,event_signal):
    print(f"Watching {file_path}")
    while not event_signal.is_set():
        print("Watching file ...")
        time.sleep(2)
    print(f"Watch thread_id : {thread_id} exiting clearly ...")

# file_path =
t1_Manually= threading.Thread(target=watch_file, args=(1,"a.txt",stop_signal_1,))
t2_Manually= threading.Thread(target=watch_file, args=(2,"b.txt",stop_signal_2,))
t3_Manually= threading.Thread(target=watch_file, args=(3,"c.txt",stop_signal_3,))


t1_Manually.start()
t2_Manually.start()
t3_Manually.start()

time.sleep(5)
stop_signal_1.set() ## Event value to be true

time.sleep(5)
stop_signal_2.set() ## Event value to be true

time.sleep(5)
stop_signal_3.set() ## Event value to be true

t1_Manually.join()
t2_Manually.join()
t3_Manually.join()

time.sleep(5)

print("Main Program exited clearly.")
```

```python
stop_event = {"thread_1":False, "thread_2":False, "thread_3":False}

def watch_file(thread_id,file_path,thread_name):
    print(f"Watching {file_path}")
    while not stop_event[thread_name]:
        print("Watching file ...")
        time.sleep(2)
    print(f"Watch thread_id : {thread_id} exiting clearly ...")

# file_path =
t1_Manually= threading.Thread(target=watch_file, args=(1,"a.txt","thread_1",))
t2_Manually= threading.Thread(target=watch_file, args=(2,"b.txt","thread_2",))
t3_Manually= threading.Thread(target=watch_file, args=(3,"c.txt","thread_3",))


t1_Manually.start()
t2_Manually.start()
t3_Manually.start()

time.sleep(5)
stop_event["thread_1"] = True  ## Event value to be true

time.sleep(5)
stop_event["thread_2"]= True  ## Event value to be true

time.sleep(5)
stop_event["thread_3"] = True  ## Event value to be true

t1_Manually.join()
t2_Manually.join()
t3_Manually.join()

time.sleep(5)

print("Main Program exited clearly.")
```

```python
stop_event = {"thread_1":False, "thread_2":False, "thread_3":False}
stop_event_lock = threading.Lock()


def watch_file(thread_id,file_path,thread_name):
    print(f"Watching {file_path}")
    while True:
        with stop_event_lock:
            if stop_event[thread_name]:
                break
            print("Watching file ...")
            time.sleep(2)
    print(f"Watch thread_id : {thread_id} exiting clearly ...")

# file_path =
t1_Manually= threading.Thread(target=watch_file, args=(1,"a.txt","thread_1",))
t2_Manually= threading.Thread(target=watch_file, args=(2,"b.txt","thread_2",))
t3_Manually= threading.Thread(target=watch_file, args=(3,"c.txt","thread_3",))


t1_Manually.start()
t2_Manually.start()
t3_Manually.start()

time.sleep(5)
with stop_event_lock:
    stop_event["thread_1"] = True  ## Event value to be true

time.sleep(5)
with stop_event_lock:
    stop_event["thread_2"]= True  ## Event value to be true

time.sleep(5)
with stop_event_lock:
    stop_event["thread_3"] = True  ## Event value to be true

t1_Manually.join()
t2_Manually.join()
t3_Manually.join()

time.sleep(5)

print("Main Program exited clearly.")
```

# ThreadPoolExecutor

```python
from concurrent.futures import ThreadPoolExecutor 
```

```python
# start()
# join()

a=[(1,"a.txt","thread_1"),(2,"b.txt","thread_2"),(3,"c.txt","thread_3"),(4,"d.txt","thread_4")]
def watch_file(input_tuple):
    thread_id, file_path, thread_name = input_tuple
    count=0
    print(f"Watching {file_path}")
    while count<5:
        print(f"Watching file ...{thread_id}")
        time.sleep(2)
        count+=1
    print(f"Watch thread_id : {thread_id} exiting clearly ...")
    return f"Hello {thread_id}"


with ThreadPoolExecutor(max_workers = 3) as executor:
    result = list(executor.map(watch_file, a))

# watch_file, (1,"a.txt","thread_1")
# watch_file, (2,"b.txt","thread_2")
# watch_file, (3,"c.txt","thread_3")
# watch_file, (4,"d.txt","thread_4")

print("Thread executed Successfully")
```
