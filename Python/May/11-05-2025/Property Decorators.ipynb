```python
# Property Of Decorators
```

```python
class Person:
    def __init__(self, name):
        self.__name = name
        self.__name1 = name
        self.__name2 = name


    ## on the getter method we have to mention the property decorator
    @property 
    def name(self):
        return self.__name
    
    def __str__(self, new_name):
        return f"person name: {self.__name}"
    
    @name.setter
    def name(self, new_name):
        if not new_name:
            raise ValueError("Name cannot be empty")
        self.__name = new_name

    @name.deleter
    def name(self):
        del self.__name
```

```python
p_property = Person("Aswathi")
```

```python
p_property.name
```

```python
p_property.name = "Arjun"
```

```python
p_property.name
```

```python
del p_property.name
```
