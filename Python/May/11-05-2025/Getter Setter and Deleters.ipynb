```python
## getter and setter 

## getters and setters are used to access and modify private attributes of a class
```

```python
class Person:
    def __init__(self, name):
        self.__name = name   ## private arrtibutes

    def get_name(self):
        """Getter method to access the private attribute __name."""
        return self.__name
    
    def __str__(self):
        return f"Person name is {self.__name}"
    
    def set_name(self, new_name):
        """Setter method to modify the private attribute __name."""
        if not new_name:
            raise ValueError("Name cannot be empty.")
        self.__name = new_name

    def delete_name(self):
        del self.__name
```

```python
p_gsd = Person('Aswathi')
```

```python
print(p_gsd)
```

```python
p_gsd.get_name()
```

```python
p_gsd.set_name("Sherya")
```

```python
p_gsd.delete_name()
```
