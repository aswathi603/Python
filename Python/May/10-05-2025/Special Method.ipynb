```python
# 3. repr

## repr is used to get a string representation of the object that can be used for debugging

class Person:
    def __init__(self, name):
        self.name  = name ## self is the instance of the class 

    def __repr__(self):
        return f"Person {self.name}"
```

```python
p_repr = Person("Aswathi")
```

```python
print(repr(p_repr))
```

```python
class Person:
    def __init__(self, name):
        self.name  = name ## self is the instance of the class 

    def __repr__(self):
        return f"Person {self.name}"

    def __str__(self):
        return f"Person name is : {self.name}"
```

```python
p_repr_str = Person("Aswathi")
```

```python
print(p_repr_str)
```

```python
print(repr(p_repr_str))
```

```python
class Person:
    def __init__(self, name):
        self.name  = name ## self is the instance of the class 

    def __repr__(self):
        return f"Person {self.name}"

    def __str__(self):
        return f"Person name is : {self.name}"
    
    def add(self,a,b):
        return a + b
```

```python
p_repr1.add(2,3)
```

```python
## 4 . __add__  it define the behaviours of + operator

class NewAdd:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"NewAdd({self.x}, {self.y})"
    
    def __add__(self, other):
        return (self.x + other.x, self.y + other.y)
    
```

```python
p__add1 = NewAdd(2, 3)
p__add2 = NewAdd(4, 5)
```

```python
p__add1 + p__add2
```

```python
# 5. __eq__ ---> it is used to compare the objects of the class

class NewEqual:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"NewEqual({self.x}, {self.y})"

    def __eq__(self,other):
        return self.x == other.x and self.y == other.y
```

```python
p__eq1 = NewEqual(2,3)
p__eq2 = NewEqual(2,3)
```

```python
p__eq1 == p__eq2
```

```python
# 6. len(self)

class Values:
    def __init__(self,items):
        self.items = items

    def __len__(self):
        return len(self.items)
```

```python
value_special = Values([1,3,5,2,1,5])
```

```python
len(value_special)
```

```python
class Val:
    def __init__(self,items):
        self.items = items
    
    def __getitem__(self, index):
        return self.items[index]
```

```python
getitems = Val([1,2,3,4,5])
```
