```python
class Calculator:

    @staticmethod
    def add(x,y):
        print(x + y)

    @staticmethod
    def subtract(x, y):
        print(x - y)

    @staticmethod
    def multiply(x, y):
        print(x * y)

    @staticmethod
    def divide(x, y):
        if y==0: 
            print("Error Cannot be divided by zero")
        else:
            print(x / y)

    def all(self, x, y):
        print(Calculator.add(x, y))
        print("Doing Something")
        print(Calculator.multiply(x, y))
        
```

```python
cal = Calculator()
cal.all(13,23)
```
