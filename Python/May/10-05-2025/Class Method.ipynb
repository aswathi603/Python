```python
# Class Method 

class B:

    # Class Bound Variable
    a=100 ## Class Variable

    def __init__(self):
        ## Objects Bound Variable
        self.a = 100 ## Instance Variable

    def hello(self):
        print("Hello World")
```

```python
#B.hello()
## this type of is called object bound method 

# ---------------------------------------------------------------------------
# TypeError                                 Traceback (most recent call last)
# Cell In[2], line 1
# ----> 1 B.hello()

# TypeError: B.hello() missing 1 required positional argument: 'self'
```

```python
B.a  ## we are calling the method without creating a object 
# B.hello() ## this will not work because we are calling the method without creating a object
```

```python
class DummyA:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    @classmethod
    def from_fullname(cls,name):  #"Aswathi 22"
        name,age = name.split()
        return cls(name,age)
        ##cls is the main class DummyA itself 

    def print_data(self):
        print(self.name)
        print(self.age)



```

```python
##DummyA("Aswathi 22")
```

```python
obj_class_method = DummyA.from_fullname("Aswathi 22")
```

```python
obj_class_method.print_data()
```
