```python
def time_it(func):
    # time_taken = 0
    def wrapper(*args):
        start_t=time.time()
        func(*args)
        end_t=time.time()
        time_taken=end_t-start_t
        print(f"Time Taken : {time_taken}")
    return wrapper
```

```python
@time_it
def pythonAdd():
    a = list(range(10000))
    b = list(range(10000))
    a = [x+y for x,y in zip(a,b)]
```

```python
time_taken_p = pythonAdd()
```

```python
def time_it(func):
    # time_taken = 0
    def wrapper(*args):
        start_t=time.time()
        func(*args)
        end_t=time.time()
        time_taken=end_t-start_t
        print(f"Time Taken : {time_taken}")
    return wrapper
```

```python
@time_it
def numpyAdd(a,b):
    a = np.array(a)
    b = np.array(b)
    a + b
```

```python
a = list(range(10000))
b = list(range(10000))
```

```python
time_taken_n = numpyAdd(a,b)
```
