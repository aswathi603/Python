```python
words.sort(key=len)
```

```python
print(words)
```

```python
words = ["apple","kiwi","banana"]
## appple - 5
## kiwi - 4
## banana - 6
```

```python
print(sorted(words,key=len))
```

```python
words = ["apple","kiwi","banana"]
## appple - 2
## kiwi - 3
## banana - 1

##sort by last letter
```

```python
print(sorted(words,key=lambda x: x[-1]))
```

### List Compherhension 

```python
#sort expressive way to generate a new list or filter a list
```

```python
nums = [1,1,2,1,1,1,12,3,5,5,7,7,2,1,1,5,1,1,1]
```

```python
## Syntax : [expression for item in iterable if codition0]
```

```python
## ps :  keep all the ones
```

```python
new_res = []
for i in nums:
    if i == 1:
        new_res.append(i)
print(new_res)
```

```python
new_res = [i for i in nums if i == 1]
```

```python
print(new_res)
```

```python
new_res = []
for i in nums:
    new_res.append(i*i)
print(new_res)
```

```python
new_res = [i*i for i in nums ]
```

```python
print(new_res)
```

```python
nums  = [1,3,4]

## p.s if the number is even then perform the square and output should be in list

```

```python
ress = [[i*i] for i in nums if i % 2 == 0 ]
```

```python
print(ress)
```

### Dict Comprehension 

```python
## {key_expression : value_expression for item in iterable  if condiotion}
```

```python
names = ["monal","Sidh","Aswathi"]
scores = [155,632,545]
```

```python
res = {name.capitalize() : score+100 for name,score in zip(names,scores)}
```

```python
print(res)
```
