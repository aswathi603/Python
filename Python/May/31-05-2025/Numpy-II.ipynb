### time counter 

```python
import time 

start_e = time.time()
np.empty((5,5))
end_e = time.time()
total_e = end_e - start_e
print(total_e)
```

### performance Counter

```python
import time 

start_e = time.perf_counter()
np.empty((5,5))
end_e = time.perf_counter()
total_e = end_e - start_e
print(total_e)
```

```python
a = np.array([[[1],[2],[3]],[[1],[2],[3]]])
print(a.shape)
b = np.squeeze(a)
print()
print(b.shape)
print(b)
```

```python
# a.shape prints (2, 3, 1)
# So:

# Axis 0: size 2
# Axis 1: size 3 ← not squeezable
# Axis 2: size 1 ← squeezable
```

```python
a = np.array([[[1],[2],[3]],[[1],[2],[3]]])
print(a.shape)
b = np.squeeze(a,axis=2)
print()
print(b.shape)
print(b)
```

```python
# It gives you control over which dimension to apply a function, like sum, mean, squeeze, etc.

# Visual:

# For a 2D array:
# axis=0 → down the rows
# axis=1 → across the columns
```

```python
a = np.array([[1,2,3],[1,2,3],[1,2,3]])
```

```python
print(a.shape," \n",a)
```

```python
print(np.sum(a,axis=0))  ## row wise  1 + 1 + 1 = 3
```

```python
print(np.sum(a,axis=1))  ## column wise it works 1 + 2 + 3
```
