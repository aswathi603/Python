### Series Indexing and Slicing

```python
s1 = pd.Series([10,20,30,40,50,60],index=['a', 'b', 'c', 'd', 'e', 'f'],name="Scores")
print(s1)
```

```python
print(s1.values[0])  ## positional index
```

```python
print(s1.loc['a']) # indexing 

# print(s1.loc[0]) # indexing
```

```python
print(s1.iloc[0])  ## Posistional Indexing 
```

```python
print(s1.loc[['a','b']]) # indexing 
```

```python
print(s1.iloc[[0,1]])
```

```python
print(s1.iloc[0:3])  ## Positional indexing (exclusive)
```

```python
#Creating series from dict 
data = {'a': 10, 'b': 20, 'c': 30, 'd': 40, 0 :100, 1: 101, 2: 400, 6:700}
s2 = pd.Series(data ,name="SCores")
```

```python
s2.loc[0:2]  ## inclusive
```

```python
## Slicing 
    ## loc [start, end] : end is inclusive 
    ## iloc [start, end] : end is exclusive
```

## Conditional Selection

```python
## get all the rows where person age is greater than 70
## df[condition]

df[df['Age']>70]

```

```python
Senior_Passanger = df[df['Age']>70]
Senior_Passanger.head()
```

```python
## get all the females who survived 

Female_Sur=df[(df['Sex'] == 'female') & (df['Survived'] == 1)]
Female_Sur.head()
```

```python
female_Pasanger = df[(df['Sex'] == 'female') & (df['Survived'] == 1)][['Name','Age']]
female_Pasanger.head()
```

```python
## total no of survivers
survived = len(df[df['Survived']==1])
nonsurvived = len(df[df['Survived']==0])
print(survived)
Total = nonsurvived+survived
```

```python
survived/Total
```
