```python
## Reindex
sales_jan  =pd.DataFrame({
    'Sales':[250,400,300,450]
}, index=['Store_A','Store_B','Store_C','Store_D'])

sales_jan
```

```python
sales_feb  =pd.DataFrame({
    'Sales':[260,310]
}, index=['Store_A','Store_C'])

sales_feb 
```

```python
common_index = sales_jan.index
```

```python
sales_feb_aligned  = sales_feb.reindex(common_index)
 
```

```python
sales_feb_aligned
```

```python
Sales_diff= sales_feb_aligned['Sales'] - sales_jan['Sales']
```

```python
Sales_diff
```

```python
sales_feb_aligned = sales_feb.reindex(common_index,fill_value=0)
sales_feb_aligned
```

```python
Sales_diff= sales_feb_aligned['Sales'] - sales_jan['Sales']
Sales_diff
```

```python
common_index_List = list(set(list(sales_jan.index) + list(sales_feb.index)))
common_index_List
```

```python
sales_feb_new_aligned = sales_feb.reindex(common_index,fill_value=0)
sales_feb_new_aligned
```

```python
sales_jan_aligned = sales_jan.reindex(common_index,fill_value=0)
sales_jan_aligned
```

```python
to_cheack  = 'Store_C'
print(sales_feb_new_aligned.loc[to_cheack])
print(sales_jan_aligned.loc[to_cheack])
```

### Iteration in Pandas

```python
df = pd.DataFrame({
    'Name' : ['Aswathi','Sidharth','Arooshi'],
    'Age'  : [23, 27, 23],
    'City' : ['New York', 'Los Angelas', 'Manchaester']
})
```

```python
for index, row in df.iterrows():
    print(row,end='\n\n')
```

```python
for index, row in df.iterrows():
    print(f"{index} : {row['Name']} lives in {row['City']} and is {row['Age']} years old.")
```

```python
for row in df.iterrows():
    print(row)
```

```python
for row in df.itertuples():
    print(row.Name)
```

```python
df.columns
```

```python
df.index
```

```python
for col in df.columns:
    print(f" Columns : {col}")
    print(df[col].values , end='\n\n')
```

```python
## condition  based  column creation using custom logic
```

```python
df.head()
```

```python
# if >=30 , senior else younger
```

```python
values = []

for idx , row in df.iterrows():
    if row['Age'] <30:
        values.append('Young')
    else:
        values.append('Senior')

df['Category'] = values
```

```python
df.head()
```

```python
# Vectorized using apply
```

```python
df = pd.DataFrame({
    'Name' : ['Aswathi','Sidharth','Arooshi'],
    'Age'  : [23, 27, 23],
    'City' : ['New York', 'Los Angelas', 'Manchaester']
})
```

```python
df['Category'] = df['Age'].apply(lambda age: 'Young' if age < 30 else 'Senior')
```
