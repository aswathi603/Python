📄 About the Dataset

This dataset is extracted from the **UCI Machine Learning Repository** and is also known as the **"Adult" or "Census Income" dataset**.

It contains information about individuals from the 1994 Census database. The goal is to predict whether a person earns more than `$50K` a year based on attributes like age, education, workclass, etc.

📌 **Source**: [UCI Adult Dataset](https://archive.ics.uci.edu/ml/datasets/adult)

```python
# 📥 Load and Preview Data
df = pd.read_csv("adult_data.csv", header=None, skipinitialspace=True)
```

```python
# Adding column names as per the UCI repository
df.columns = [
    "age", "workclass", "fnlwgt", "education", "education-num", "marital-status",
    "occupation", "relationship", "race", "sex", "capital-gain", "capital-loss",
    "hours-per-week", "native-country", "income"
]
```

```python
# Show the first 5 rows
df.head()
```

```python
## Dataset info 
df.info()
```

```python
#Summary of Unique Values per Column
df.nunique()
```

```python
# Check for missing values
print("\nMissing values before cleaning:")
print(df.isnull().sum())
```

```python
# Drop rows with missing values
df.dropna(inplace=True)
```

errors='coerce': If any value cannot be converted, it will be replaced with NaN (Not a Number). This allows the conversion to continue without raising an error.

```python
# 2. Convert data types
# Convert 'age' and 'hours-per-week' to numeric
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['hours-per-week'] = pd.to_numeric(df['hours-per-week'], errors='coerce')
```

```python
# Convert categorical variables to category type
categorical_columns = ['workclass', 'education', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'income']
for col in categorical_columns:
    df[col] = df[col].astype('category')

print(categorical_columns)
```

```python
df = pd.get_dummies(df, columns=categorical_columns[:-1], drop_first=True)
df.head()
```

```python
# 4. Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
#  5. Reset the index
df.reset_index(drop=True, inplace=True)
```

```python
# Display the cleaned data
print("\nCleaned Data:")
print(df.head())
```

```python
# Save the cleaned data to a new CSV file
df.to_csv('cleaned_adult_data.csv', index=False)
print("\nCleaned data saved to 'cleaned_adult_data.csv'.")
```

### Interesting Textual Insights

```python
#Income Distribution 
income_count = df['income'].value_counts()
display(income_count)
```

```python
#Converting income to numeric value 
df['income_numeric'] = df['income'].map({'>50K': 1, '<=50K': 0})
```

```python
# Group By martial status and calculate the average income 
marital_income  = df.groupby('marital-status')['income_numeric'].mean()
print(marital_income)
```

```python
# Relationship between Age and Income
#Calculate average income by age
age_income = df.groupby('age')['income_numeric'].mean()
print(age_income)
```

```python
# Effect of Education on Occupation and Income
# Group by education and Occupation type , then calculate average income 
education_job_income = df.groupby(['education','occupation'])['income_numeric'].mean()
print(education_job_income)
```

```python
# Correlation Between Hours Worked Per Week and Income
# convert if you want to check it is not numeric 
df['hours-per-week'] = pd.to_numeric(df['hours-per-week'])

# Calculation correlation
correlation = df['hours-per-week'].corr(df['income_numeric'])
print("Correlation Between Hours Worked Per Week and Income: ",correlation)
#
```

```python
# Influence of Gender on Income Disparities
gender_income = df.groupby('sex')['income_numeric'].mean()
print(gender_income)
```

```python
# Relationship Between Race and Income Levels
race_income = df.groupby('race')['income_numeric'].mean()
print(race_income)
```

```python
# Geographical Trends in Income Based on State or Region
state_income = df.groupby('native-country')['income_numeric'].mean()
print(state_income)
```
